## 2. Dataset Overview & Loading

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
import plotly.graph_objects as go
from plotly.subplots import make_subplots
import warnings
warnings.filterwarnings('ignore')

# Load Dataset
df = pd.read_csv('/kaggle/input/datasets/alitaqishah/esg-greenwashing-energy-and-industrials/esg_greenwashing_energy_utilities_industrials_2010_2024.csv')

print(f"✅ Dataset Shape: {df.shape[0]:,} rows × {df.shape[1]} columns")
print(f"📅 Years Covered: {df['year'].min()} - {df['year'].max()}")
print(f"🏢 Companies: {df['company'].nunique()}")
print(f"📍 Sectors: {df['sector'].unique()}")
```

```python
df.head(10)
```

```python
df.describe()
```

## 3. Data Quality & Preprocessing

```python
# Missing Values Analysis
missing = df.isnull().sum()
missing_pct = (missing / len(df) * 100).round(2)
missing_df = pd.DataFrame({'Missing Count': missing, 'Missing %': missing_pct})
print(missing_df[missing_df['Missing Count'] > 0])

# Handle missing values
df['yoy_scope1_change_pct'] = df['yoy_scope1_change_pct'].fillna(0)
df.fillna({'cdp_climate_score': 'Not Rated'}, inplace=True)
```

## 4. Exploratory Data Analysis (EDA)


### 4.1 Distribution of Key Metrics

```python
fig, axes = plt.subplots(2, 2, figsize=(16, 12))

sns.histplot(data=df, x='revenue_usd_bn', bins=30, ax=axes[0,0], kde=True)
axes[0,0].set_title('Revenue Distribution')

sns.histplot(data=df, x='total_s1_s2_mt_co2e', bins=30, ax=axes[0,1], kde=True)
axes[0,1].set_title('Total Scope 1+2 Emissions')

sns.histplot(data=df, x='esg_score_0_100', bins=20, ax=axes[1,0], kde=True)
axes[1,0].set_title('ESG Score Distribution')

sns.histplot(data=df, x='carbon_intensity_tco2e_per_musd', bins=30, ax=axes[1,1], kde=True)
axes[1,1].set_title('Carbon Intensity Distribution')

plt.tight_layout()
plt.show()
```

## 5. Emissions Analysis & Carbon Intensity Trends

```python
plt.figure(figsize=(14, 8))
trend = df.groupby(['year', 'sector'])['carbon_intensity_tco2e_per_musd'].mean().reset_index()

for sector in trend['sector'].unique():
    data = trend[trend['sector'] == sector]
    plt.plot(data['year'], data['carbon_intensity_tco2e_per_musd'], 
             marker='o', linewidth=2.5, label=sector)

plt.title('Carbon Intensity Trend by Sector (2010-2024)', fontsize=16)
plt.xlabel('Year')
plt.ylabel('tCO₂e per Million USD Revenue')
plt.legend(title='Sector', fontsize=12)
plt.grid(True)
plt.show()
```

## 6. Greenwashing Detection & Patterns

```python
plt.figure(figsize=(10, 6))
gw_rate = df.groupby('sector')['greenwashing_flag'].mean() * 100
gw_rate.plot(kind='bar', color=sns.color_palette("Blues_r", 3))
plt.title('Greenwashing Rate by Sector (%)', fontsize=16)
plt.ylabel('Percentage (%)')
plt.xlabel('Sector')
plt.xticks(rotation=0)
for i, v in enumerate(gw_rate):
    plt.text(i, v + 0.5, f'{v:.1f}%', ha='center')
plt.show()
```

## 7. ESG Scores, CDP & Net-Zero Commitments

```python
plt.figure(figsize=(14, 8))
esg_trend = df.groupby(['year', 'sector'])['esg_score_0_100'].mean().reset_index()

for sector in esg_trend['sector'].unique():
    data = esg_trend[esg_trend['sector'] == sector]
    plt.plot(data['year'], data['esg_score_0_100'], 
             marker='o', linewidth=2.5, label=sector)

plt.title('Average ESG Score Trend by Sector (2010-2024)', fontsize=16)
plt.xlabel('Year')
plt.ylabel('ESG Score (0-100)')
plt.legend(title='Sector')
plt.grid(True)
plt.show()
```

## 8. Sector & Country Comparative Analysis

```python
fig, axes = plt.subplots(1, 2, figsize=(16, 7))

sns.boxplot(data=df, x='sector', y='carbon_intensity_tco2e_per_musd', ax=axes[0])
axes[0].set_title('Carbon Intensity by Sector')

sns.boxplot(data=df, x='sector', y='esg_score_0_100', ax=axes[1])
axes[1].set_title('ESG Score by Sector')

plt.tight_layout()
plt.show()
```

## 9. Time Series & Company-Level Deep Dives

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Filter data for selected companies
companies = ['ExxonMobil', 'NextEra Energy']
comp_df = df[df['company'].isin(companies)]

# Set style
plt.style.use('seaborn-v0_8-whitegrid')
sns.set_palette("Set2")

# Create subplots
fig, axes = plt.subplots(2, 1, figsize=(14, 12))

# Plot 1: ESG Score Comparison
for company in companies:
    company_data = comp_df[comp_df['company'] == company]
    axes[0].plot(company_data['year'], company_data['esg_score_0_100'], 
                 marker='o', linewidth=2.5, label=company)

axes[0].set_title('ESG Score Comparison: ExxonMobil vs NextEra Energy', fontsize=16, pad=15)
axes[0].set_xlabel('Year', fontsize=12)
axes[0].set_ylabel('ESG Score (0-100)', fontsize=12)
axes[0].legend(title='Company', fontsize=11)
axes[0].grid(True)

# Plot 2: Total Scope 1+2 Emissions Comparison
for company in companies:
    company_data = comp_df[comp_df['company'] == company]
    axes[1].plot(company_data['year'], company_data['total_s1_s2_mt_co2e'], 
                 marker='o', linewidth=2.5, label=company)

axes[1].set_title('Total Scope 1+2 Emissions Comparison: ExxonMobil vs NextEra Energy', 
                  fontsize=16, pad=15)
axes[1].set_xlabel('Year', fontsize=12)
axes[1].set_ylabel('Emissions (Mt CO₂e)', fontsize=12)
axes[1].legend(title='Company', fontsize=11)
axes[1].grid(True)

plt.tight_layout()
plt.show()
```

## 10. Correlation, Statistical Testing & Insights

```python
plt.figure(figsize=(12, 10))
numeric_cols = ['revenue_usd_bn', 'scope1_emissions_mt_co2e', 'total_s1_s2_mt_co2e',
                'carbon_intensity_tco2e_per_musd', 'esg_score_0_100', 
                'greenwashing_flag', 'scope3_emissions_mt_co2e']

corr = df[numeric_cols].corr()
sns.heatmap(corr, annot=True, cmap='coolwarm', center=0, fmt='.2f', linewidths=0.5)
plt.title('Correlation Matrix: Emissions vs ESG Metrics', fontsize=16)
plt.show()
```

## 11. Predictive Modeling & Feature Importance

```python
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import classification_report, confusion_matrix

features = ['revenue_usd_bn', 'total_s1_s2_mt_co2e', 'carbon_intensity_tco2e_per_musd',
            'esg_score_0_100', 'scope1_emissions_mt_co2e', 'scope3_emissions_mt_co2e']

X = df[features].fillna(0)
y = df['greenwashing_flag']

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=42)

model = RandomForestClassifier(n_estimators=300, random_state=42)
model.fit(X_train, y_train)

pred = model.predict(X_test)
print(classification_report(y_test, pred))
```

## 12. Conclusions, Recommendations & Limitations

### 🎯 Key Conclusions
1. **Greenwashing exists** — ESG scores improved faster than actual emission reductions.
2. **Utilities** lead in real decarbonization.
3. **Energy majors** show the highest greenwashing signals.
4. Absolute emissions remain stubbornly high despite rising ESG ratings.

### 💡 Recommendations
- **Investors**: Prioritize absolute emission reduction over ESG scores.
- **Regulators**: Mandate standardized, third-party verified Scope 3 reporting.
- **Companies**: Align public claims with verifiable, science-based targets.

**Limitations**: Self-reported data, limited to 21 large companies.
