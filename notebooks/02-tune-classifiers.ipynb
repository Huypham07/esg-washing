## Tổng hợp best params

`03-train-classifiers.ipynb` **tự nạp** `best_params_<task>.json` → KHÔNG cần sửa `config/train.yml` tay. Best params giờ gồm `warmup_ratio` (không còn `epochs`).

```python
import json, pandas as pd
rows = []
for t in TASKS:
    p = ROOT / f"outputs/models/{OUTDIR[t]}/best_params_{t}.json"
    if p.exists():
        rows.append({"task": t, **json.load(open(p))})
    else:
        print("chưa có:", p)
pd.DataFrame(rows)
```
